### 합성곱(Convolution)
- 인공 신경망은 처음에 가중치들과 절편을 랜덤하게 초기화
    - 에포크 반복하면서 경사 하강법 알고리즘으로 손실이 낮아지는 최적의 가중치와 절편 찾아감 -> 모델 훈련
- 합성곱은 밀집층처럼 입력 데이터 전체에 가중치 적용 X
  - **일부에만 가중치 곱함**
- 직관적으로 다음과 같이 이해할 수 있음
  - 10개의 입력값이 있고, 합성곱 층의 뉴런은 3개의 특성을 입력으로 받음
  - 그리고 한 칸씩 내려가면서 3개씩 입력으로 집어넣어가면서 출력 만듦
    - 1~3번째, 2~4번째, 3~5번째, ...
  - **이때 어떤 합성곱에서 사용된 가중치가 다른 합성곱에서도 동일하게 사용**되는 게 중요한 특징
    - $a_1 \times w_1 + a_2 \times w_2 + a_3 \times w_3$
    - $a_2 \times w_1 + a_3 \times w_2 + a_4 \times w_3$
    - ...
    - $a_8 \times w_1 + a_9 \times w_2 + a_{10} \times w_3$
  - 이렇게 모두 같은 뉴런에서 8개의 출력이 만들어짐
- 합성곱층의 뉴런에 있는 가중치 개수 -> 하이퍼파라미터

- 합성곱의 장점 -> 2차원 입력에도 적용 가능
  - 입력이 2차원 배열 -> 필터도 2차원
  - 왼쪽 위에서 오른쪽 맨 아래까지 이동하면서 출력 만들면 됨
  - 계산식은 밀집층과 똑같이 입력과 가중치 행과 열 맞춰서 곱하고 모두 더하면 됨
- 합성곱 계산을 통해 얻은 출력을 **특성 맵**(feature map)이라 함
  - 일반적으로 활성화 함수를 적용한 후
- 합성곱 층에서도 **여러 개의 필터**를 사용할 수 있음
  - **여러 필터를 사용한 특성 맵은 순서대로 쌓임**
  - ex) (2, 2) 크기의 특성 맵을 쌓으면 3차원 배열

##### 필터(커널)
- 기존에 배운 신경망 층
  - 뉴런이 세로로 길게 늘어 서있고, 서로 조밀하게 연결됨
- 합성곱에서는 뉴런이 **입력 위를 이동하면서 출력을 만듦** -> 기존과 같이 표현하기 어려움
  - 뉴런이라고 부르기에도 애매함
- 합성곱 신경망(convolutional neural network, CNN)에서는 완전 연결 신경망과 달리 뉴런을 필터(또는 커널)이라고 함
  - 뉴런 = 필터 = 커널 이라고 생각해도 됨
- 크기는 하이퍼파라미터 -> 직접 여러 값 시도해봐야 함
  - 보통 (3, 3), (5, 5) 많이 사용

```python
# 케라스 합성곱 층
import keras
# Conv2D: 입력 왼쪽->오른쪽 / 위->아래 이동하는 합성곱
keras.layers.Conv2D(10, kernel_size=(3, 3), activation='relu') # 필터 개수, 커널 사이즈, 활성화 함수
```

### 합성곱 신경망
- 1개 이상의 합성곱 층을 사용한 인공 신경망
  - 꼭 합성곱 층만 사용할 필요는 X

### 패딩, 스트라이드
- (4, 4) 입력에서 (3, 3) 커널로 (4, 4) 출력 만들기 -> 더 큰 입력을 합성곱하는 척 해야 함
- 입력 배열 주위를 가상의 원소로 채우는 것 -> **패딩**
  - 실제 입력값이 아니므로 0으로 채움
  - 순전히 커널 계산을 도와주기만 함 -> 계산값에는 영향 X
- **입력과 특성 맵의 크기를 동일하게 하기 위해 입력 주위에 0으로 패딩**하는 것 -> **세임 패딩**(same padding)
  - 일반적으로 많이 사용
- 패딩 없이 순수한 입력 배열에서만 합성곱을 해 특성 맵 만듦 -> **밸리드 패딩**(valid padding)
  - 특성 맵의 크기가 줄어들 수밖에 없음
- 패딩을 사용하는 이유
  - 패딩이 없으면 구석에 있는 원소들은 커널과 계산되는 횟수가 적음 -> 중요한 정보가 특성 맵으로 전달되지 않을 수 있음
- 커널을 움직이는 크기를 **스트라이드**(stride)라 함
  - 스트라이드가 클 수록 특성 맵의 크기는 작아짐

```python
keras.layers.Conv2D(10, kernel_size=(3, 3), activation='relu', padding='same', strides=1) # Same Padding, 스트라이드 1
```

### 풀링(Pooling)
- 합성곱 층에서 만든 특성 맵의 개수는 유지하면서 가로세로 크기 줄이는 역할
- 예시
  - (4, 4) 입력을 (3, 3) 크기의 커널 3개로 (2, 2) 특성맵 3개를 만듦 -> (2, 2, 3)
  - 이 특성맵에 풀링을 적용하면 마지막 차원인 개수는 그대로 유지 & 너비와 높이만 줄임 -> (1, 1, 3)
- **방식**
  - 합성곱처럼 입력 위를 지나가면서 영역을 찍는데, 가중치는 없음
  - 영역 내에서 가장 큰 값을 고르거나, 평균값을 계산함
    - 각각 **최대 풀링**(max pooling), **평균 풀링**(average pooling)
    - 주로 최대 풀링 사용 -> 평균 풀링은 특성 맵에 있는 중요한 정보를 평균하여 희석시킬 수 있음
- 풀링은 합성곱 층과 구분되기 때문에 **풀링 층**이라고 부름
- 풀링 층의 출력도 똑같이 특성 맵
- 풀링과 합성곱의 큰 차이
  - 합성곱은 커널이 보통 한 칸씩 이동하므로 겹치는 부분이 있음
  - **풀링에서는 겹치지 않고 이동**
    - ex) 풀링 크기가 (2,2)면 스트라이드가 2, (3,3)이면 스트라이드가 3
